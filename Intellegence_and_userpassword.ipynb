```python
try:
  score = int(input("Please enter your score to determine your intelligence: "))
  if score >= 20:
    print("Smart")
  else:
    print("Dumb")
except ValueError:
  print("You are really dumber than Patrick Star")
```

```python
# Predefined credentials
correct_username = "admin"
correct_password = "secret123"

# Configure maximum attempts
max_attempts = 3
attempts = 0
logged_in = False

while attempts < max_attempts:
    username_input = input("Enter username: ")
    password_input = input("Enter password: ")

    if username_input == correct_username and password_input == correct_password:
        print("Login successful! Welcome.\n")
        logged_in = True
        break
    else:
        attempts += 1
        remaining = max_attempts - attempts
        print(f"Invalid username or password. Remaining attempts: {remaining}\n")

if not logged_in:
    print("Too many failed attempts. Access locked.")
```
