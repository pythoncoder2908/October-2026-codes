```python
try:
  game_score = int(input("Enter your game score to determine your capabilities: "))
  if game_score < 0:
    raise ValueError("Score cannot be negative")
  if game_score >= 80:
    print("Professional")
  elif game_score >= 45:
    print("Mediocre")
  else:
    print("Beginner")
except ValueError as e:
     print(e)
```
